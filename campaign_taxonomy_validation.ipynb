```python
import pandas as pd

print("Libraries Loaded Successfully")
```

```python
data = {
    "campaign_name": [
        "FB_SOC_Philips_NAM_US_CV_USD_820804315294108",
        "GOOG_SEA_Philips_EMEA_UK_TR_GBP_920804315294109",
        "DV360_DSP_Philips_APAC_IN_AW_INR_720804315294110",
        "Facebook_Philips_US"
    ]
}

df = pd.DataFrame(data)

df
```

```python
def validate_campaign_name(campaign):

    parts = campaign.split("_")

    if len(parts) == 8:
        return "VALID"

    return "INVALID"
```

```python
df["taxonomy_status"] = df["campaign_name"].apply(validate_campaign_name)

df
```

```python
df["taxonomy_status"].value_counts()
```

```python
import matplotlib.pyplot as plt

df["taxonomy_status"].value_counts().plot(
    kind="bar",
    color=["green", "red"]
)

plt.title("Campaign Taxonomy Validation Results")
plt.xlabel("Status")
plt.ylabel("Count")

plt.show()
```
